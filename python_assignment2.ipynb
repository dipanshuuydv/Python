1. Ask the user for a string and check whether it is a palindrome or not. A  palindrome “madam”, “ is a string which is same when we read it forward & backward. Eg - racecar” etc.

```python
txt = "racecar"

ntxt = ""

for i in range(len(txt)-1,-1,-1):
    ntxt = ntxt + txt[i]

if ntxt == txt:
    print(f"Yes {txt} is Palindrome")
else:
    print(f"No, {txt} is not Palindrome")
```

2. Given a list of integers compute the average of all numbers in the list.

```python

num = [45,85,47,65,49,16]
avg = 0
sum = 0
for i in range(len(num)):
    sum += num[i]

avg = sum/(len(num)+1)

print(f"Average is {round(avg, 2)}")
```

4. Given a tuple of integers, create: 
• A tuple of all even numbers 
• A tuple of all odd numbers

```python
numbers = (10, 15, 20, 25, 30, 35, 40, 45)
even = ()
odd = ()


for i in numbers :
    if i % 2 == 0 : 
        even += (i,)
    else : 
        odd += (i,)

print(f"Given Tuple is : {numbers}")
print(f"Even Tuple are : {even}")
print(f"Odd Tuple are : {odd}")
```

5. Create a dictionary where: 
• Keys = student names 
• Values = marks (integer) 

        Write a menu-based program where user presses a key (ʼAʼ, ‘Bʼ, ‘Cʼ, ‘Dʼ) depending on the operation they want to perform on the dictionary: 

    1.  A - Add a student
    2.  B - Update marks
    3.  C - Search for a student
    4.  D - Display all students and marks

```python
students = {}

def add_student() :
    sname = input("Enter the Name of Student : ").capitalize()
    marks = int(input(f"Enter the Marks of {sname} : "))
    students[sname] = marks

def update_marks() :
    sname = input("Enter the Name of Student to update Marks : ").capitalize()
    if sname in students:
        marks = int(input(f"Enter the marks of {sname} to Update : "))
        students[sname] = marks
        print(f"Marks updated for {sname} to {marks}")

def Search_for_students() :
    sname = input(f"Enter Student Name to Search : ").capitalize()
    if sname in students:
        print(students)
    else :
        print(f"Student not found with name {sname}.")

def display_all() :
    print(students)

while True:
    print("===============================================")
    print("A. Add a Student.")
    print("B. Update Marks.")
    print("C. Search for a Student.")
    print("D. Display all Students with their Marks.")
    print("E. Exit.")
    ch = input("Choose an option A to E : ").upper()

    if ch == 'A' :
        add_student()
    elif ch == 'B' :
        update_marks()
    elif ch == 'C' :
        Search_for_students()
    elif ch == 'D' :
        display_all()
    elif ch == 'E' :
        break
    else :
        print("Enter valid Option A to E ")

    print("Thanks for Using !!")
```

6. Given a list of words:

words = ["apple", "banana", "kiwi", "cherry", "mango"]

Create a dictionary that maps each word to its length. 
Example:

{"apple": 5, "banana": 6, "kiwi": 4, ...}

```python
fruits = ["apple", "banana", "kiwi", "cherry", "mango"]
fruit_list = {}

for fruit in fruits :
    fruit_list[fruit] = len(fruit)

print(fruit_list)
```

8. Write a program to check whether two lists share no common elements.

>> share no common elements list1 = [1, 2, 3, 4] list2 = [5, 6, 7, 8]

>> share common elements list1 = [1, 2, 3] list2 = [3, 4]

[ Hint - use sets ]

9. Given a list, print all elements that appear more than once in the list.

>> [ Hint- use sets]

```python
items = [1, 2, 3, 2, 4, 5, 3, 2, 6, 1]

unique = set()
duplicate = set()

for item in items :
    if item in unique:
        duplicate.add(item)
    else :
        unique.add(item)
        
print(f"The elements which appear more than once in a list are: {duplicate}")        
```

10. Ask the user for a string and print:

>> All unique characters

>> The count of unique characters

```python
from itertools import count


str = input("Enter any String : ")
unique = set()
for s in str:
    if s in unique:
        continue
    else:
        unique.add(s)
        
print(f"The Unique Characters in {str} are : {unique}")
print(f"Total unique Characters are : {len(unique)} ")
```
