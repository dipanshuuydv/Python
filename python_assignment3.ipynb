Add methods to deposit, withdraw, and check balance.

```python
class BankAccount :
    def __init__ (self, acNo, owner_name, balance = 0):
        self.acNo = acNo
        self.owner_name = owner_name
        self.balance = balance

    def withdraw(self, amt):
        if amt > self.balance:
            print("Low Balance !!")
        else : 
            self.balance -= amt
            print(f"{amt} is withdrawn your new balance is {self.balance}")
    
    def deposit(self, amt):
        self.balance += amt
        print(f"Deposited {amt} your new balance is {self.balance}")

    def checkBalance(self) :
        print(f"Your current Balance is {self.balance}")

acc1 = BankAccount("101", "Dipanshu", 20000)
acc1.checkBalance()
acc1.deposit(2500)
acc1.withdraw(15000)
```

```python
class Message:
    def __init__(self, sender, text):
        self.sender = sender
        self.text = text

    def __str__(self):
        return f"{self.sender.name}: {self.text}"


class User:
    def __init__(self, name):
        self.name = name

    def send_message(self, chatroom, text):
        chatroom.add_message(self, text)


class ChatRoom:
    def __init__(self, name):
        self.name = name
        self.users = []
        self.messages = []

    def join(self, user):
        if user in self.users:
            print(f"{user.name} pehle se room mein hai.")
        else:
            self.users.append(user)
            print(f"{user.name} ne '{self.name}' join kiya.")

    def leave(self, user):
        if user in self.users:
            self.users.remove(user)
            print(f"{user.name} ne '{self.name}' chhod diya.")
        else:
            print(f"{user.name} room mein hai hi nahi.")

    def add_message(self, user, text):
        if user not in self.users:
            print(f"{user.name} room mein nahi hai, message nahi bhej sakta.")
            return
        msg = Message(user, text)
        self.messages.append(msg)

    def show_history(self):
        print(f"--- {self.name} ki chat history ---")
        if not self.messages:
            print("Abhi koi message nahi hai.")
        for msg in self.messages:
            print(msg)


# ---------- Test ----------
room = ChatRoom("Friends")

rahul = User("Rahul")
priya = User("Priya")

room.join(rahul)
room.join(priya)

rahul.send_message(room, "Hello Priya!")
priya.send_message(room, "Hi Rahul, kaise ho?")

room.show_history()

room.leave(priya)
priya.send_message(room, "Mujhe koi nahi sunega")   # error message aayega
```
