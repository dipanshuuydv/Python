### Area of Circle

```python
class Circle :
    pi = 0

    def __init__ (self, radius):
        self.radius = radius 

    @staticmethod
    def pi_value():
        return 3.14

    def area(self):
        return self.pi_value() * self.radius ** 2

c = Circle(10)
print(c.area())
```

```python
class Employee:
    company = "TCS"

    def __init__ (self, name,salary = 0):
        self.name = name
        self.salary = salary
        print(f"I am {self.name} and my salary is {self.salary}")

    @classmethod
    def change_company(cls,company):
        cls.company = company


    
e1 = Employee("Rina", 50000)
e2 = Employee("Rahul", 70000)

Employee.change_company("Wipro")
print(e1.company)
print(e2.company)
```
