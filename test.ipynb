```python
info = [
    ("Alice", "Math"),
    ("Bob", "Science"),
    ("Alice", "Science"),
    ("Charlie", "Math"),
    ("Bob", "Math"),
    ("Alice", "English"),
    ("Charlie", "English"),
    ("Diana", "History"),
    ("Evan", "Math"),
    ("Diana", "Science"),
]

courses = set()
students = {}

for name, course in info :
    if name not in students :
        students.update({name:set()})
        students[name].add(course)
    else:
        students[name].add(course)
         
    
# print(courses)
print(students)
```
