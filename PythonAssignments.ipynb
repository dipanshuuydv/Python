    • If salary < 30,000 → 5% 
    • If salary is 30,000–70,000 → 15%

2. Write a function that takes two integers  and  and prints all even numbers between them (inclusive)

```python
def printEven(a, b):
    for i in range(a, b):
        if i%2 == 0:
            print(i, end = " ")

printEven(2,20)
```
