2. Write a function that takes two integers  and  and prints all even numbers between them (inclusive)

```python
def printEven(a, b):
    for i in range(a, b):
        if i%2 == 0:
            print(i, end = " ")

printEven(2,20)
```

5. Write a function to return the sum of digits of a number, n.

```python
def sumOfDigits(num):
    n = num
    sum = 0
    r = 0
    while n > 0:
        r = n%10;
        sum = sum + r;
        n = n//10

    return sum;

print(sumOfDigits(4547835))
```

6. Write a program to print all numbers from 1 to 100 that are divisible by both 3 and 5.

7. Design a program to continuously input a number  from user & print if it is positive or negative until the user enters “Quit”

```python
while True:
    s = input("Enter a Number or Quit")
    
    if s =="Quit":
        break;

    n = float(s)
    if n >= 0:
        print(f"{n} is Positive")
    else:
        print(f"{n} is Negative")
```
